# Random Forest & Neural Network Classifier — Spheroid Viability Prediction



This notebook trains a **Random Forest (RF)** and a **Neural Network (NN)** to classify
tumour spheroid viability (Low viability / High viability) from morphological and texture features extracted
at multiple time-points (day 0, 1, 2) after photothermal treatment (PTT).

**Pipeline:**
1. Data loading and filtering
2. Feature extraction: morphology + texture (temperature excluded to prevent leakage)
3. Feature engineering: temporal dynamics (growth rates, grey-value changes, etc.)
4. Feature selection: SelectKBest / all features (configurable)
5. 5-fold stratified cross-validation (RF)
6. Overfitting analysis (RF)
7. Final model training with optional SMOTE class balancing
8. Comprehensive visualisations (confusion matrix, ROC, feature importance)
9. Bayesian MCMC IT₅₀ estimation from temperature vs predicted viability
10. Neural network training, 5-fold CV, and RF vs NN comparison
11. Violin plots and metrics figures
12. CSV export of out-of-fold predictions

## 1. Load and Prepare Data

```python
print("\n[1/7] Loading data...")
OUTPUT_DIR = Path(TRAIN_CSV).parent   # figures and CSVs are saved next to the input file

def _filter_data(df, label=''):
    """Keep only rows with a valid Label_day2 and a valid temperature value."""
    # Must have a day-2 label (target variable)
    df = df[df['Label_day2'].notna()].copy()

    # Must have a temperature measurement — rows without it cannot appear in
    # section 8a plots and add noise, so exclude them up front
    if MAX_TEMP_COL in df.columns:
        before = len(df)
        df = df[df[MAX_TEMP_COL].notna()].copy()
        dropped = before - len(df)
        if label:
            print(f"  {label} dropped (no temperature): {dropped} rows")
    else:
        if label:
            print(f"  WARNING: '{MAX_TEMP_COL}' not found in {label} — temperature filter skipped")

    # Reset to a clean 0-based integer index so loc/iloc are always consistent
    # and duplicate original IDs never cause row multiplication.
    df = df.reset_index(drop=True)
    return df

# Load training data
df_train_raw = pd.read_csv(TRAIN_CSV, sep=';', index_col=0)
df_complete  = _filter_data(df_train_raw, label='Train')

if CROSS_DATASET_MODE:
    # Load a completely separate test file — no rows shared with the training set
    df_test_raw      = pd.read_csv(TEST_CSV, sep=',', index_col=0)
    df_complete_test = _filter_data(df_test_raw, label='Test')
    print(f"  Mode            : CROSS-DATASET")
    print(f"  Train file      : {Path(TRAIN_CSV).name}")
    print(f"  Test  file      : {Path(TEST_CSV).name}")
    print(f"  Train samples   : {len(df_complete)}  "
          f"(Alive={( df_complete['Label_day2']=='Alive').sum()}, Dead={( df_complete['Label_day2']=='Dead').sum()})")
    print(f"  Test  samples   : {len(df_complete_test)}  "
          f"(Alive={(df_complete_test['Label_day2']=='Alive').sum()}, Dead={(df_complete_test['Label_day2']=='Dead').sum()})")
else:
    # Internal split — test set will be carved out in section 7
    df_complete_test = None
    print(f"  Mode         : internal 75/25 split")
    print(f"  Total samples: {len(df_complete)}")
    print(f"  Alive: {(df_complete['Label_day2']=='Alive').sum()}")
    print(f"  Dead:  {(df_complete['Label_day2']=='Dead').sum()}")
    display(df_complete)
```

## 2. Feature Extraction (No Temperature)

```python
print("\n[2/7] Extracting features...")

# Morphological and texture features to include — temperature columns are deliberately excluded
# to avoid data leakage (temperature is used as a predictor variable later, not a feature)
include_patterns = [
    'Area (pix2)', 'Circularity', 'Solidity', 'Perimeter (pix)', 'Aspect ratio',
    'Mean grey value', 'Energy', 'Homogeneity', 'Correlation', 'Equivalent Diameter (pix)'
]

# Build the day suffixes to retain based on DAYS_TO_USE and INCLUDE_DELTA
_day_suffixes = [f'_{d}' for d in DAYS_TO_USE]
if INCLUDE_DELTA:
    _day_suffixes.append('_delta')   # pre-computed day-to-day difference columns in the CSV


def _apply_nan_strategy(X, strategy):
    """Handle NaN values according to the chosen strategy."""

    if strategy == 'interpolate':
        # For each feature, treat day0/day1/day2 as time steps and interpolate per row.
        # This is biologically meaningful: a missing day1 value is estimated from day0 and day2
        # of the same spheroid rather than from other spheroids.
        day_order = ['_day0', '_day1', '_day2']
        base_groups = {}
        for col in X.columns:
            for suffix in day_order:
                if suffix in col:
                    base = col.replace(suffix, '')
                    base_groups.setdefault(base, {})[suffix] = col
                    break

        X = X.copy()
        for base, suffix_col_map in base_groups.items():
            ordered_cols = [suffix_col_map[s] for s in day_order if s in suffix_col_map]
            if len(ordered_cols) < 2:
                continue   # need at least 2 time points to interpolate
            # axis=1 interpolates across columns (time axis) for each row independently
            interpolated = X[ordered_cols].interpolate(axis=1, limit_direction='both')
            X[ordered_cols] = interpolated

        # Fallback for rows where all days are NaN — use column median
        for col in X.columns:
            if X[col].isna().any():
                X[col] = X[col].fillna(X[col].median())

    elif strategy == 'drop':
        # Remove any row that still has a NaN — simplest approach, loses data
        rows_before = len(X)
        X = X.dropna()
        rows_after = len(X)
        print(f"    'drop' strategy: removed {rows_before - rows_after} rows with NaN "
              f"({rows_before} → {rows_after})")

    else:  # 'median' — replace each NaN with the column median (default)
        X = X.copy()
        for col in X.columns:
            if X[col].isna().any():
                X[col] = X[col].fillna(X[col].median())

    return X


def extract_features(df, label=''):
    """Select numerical morphology columns matching the configured days, then handle NaNs."""
    numerical_features = []
    for col in df.columns:
        if any(pattern in col for pattern in include_patterns):
            if any(suffix in col for suffix in _day_suffixes):
                if df[col].dtype in ['float64', 'int64', 'float32', 'int32']:
                    numerical_features.append(col)

    X = df[numerical_features].copy()
    nan_count = X.isna().sum().sum()
    X = _apply_nan_strategy(X, NAN_STRATEGY)

    if label:
        print(f"  {label} base features : {len(numerical_features)}")
        print(f"  {label} NaN cells     : {nan_count}  →  strategy: '{NAN_STRATEGY}'")
        print(f"  {label} rows after NaN: {len(X)}")
    return X, numerical_features


print(f"  Days used   : {DAYS_TO_USE}  |  delta: {INCLUDE_DELTA}")
X_features, numerical_features = extract_features(df_complete, label='Train')

# When NAN_STRATEGY='drop', rows are removed from X_features.
# Align df_complete to the surviving rows so y and all later index lookups stay consistent.
if len(X_features) < len(df_complete):
    df_complete = df_complete.loc[X_features.index]
    print(f"  df_complete aligned to X_features: {len(df_complete)} rows")

if CROSS_DATASET_MODE:
    X_features_test, _ = extract_features(df_complete_test, label='Test')
    if len(X_features_test) < len(df_complete_test):
        df_complete_test = df_complete_test.loc[X_features_test.index]
        print(f"  df_complete_test aligned to X_features_test: {len(df_complete_test)} rows")
else:
    X_features_test = None
```

## 3. Feature Engineering

```python
print("\n[3/7] Engineering features...")

def engineer_features(X_features):
    """
    Create derived features that capture temporal dynamics between days.
    All computations are guarded by existence checks so the function works
    correctly even if only a subset of days was loaded.
    """
    X = X_features.copy()

    # Identify day-specific columns for each relevant feature
    area_day0  = [c for c in X.columns if 'Area' in c and 'day0' in c]
    area_day1  = [c for c in X.columns if 'Area' in c and 'day1' in c]
    area_day2  = [c for c in X.columns if 'Area' in c and 'day2' in c]
    grey_day0  = [c for c in X.columns if 'Mean grey value' in c and 'day0' in c]
    grey_day1  = [c for c in X.columns if 'Mean grey value' in c and 'day1' in c]
    grey_day2  = [c for c in X.columns if 'Mean grey value' in c and 'day2' in c]
    homog_day2 = [c for c in X.columns if 'Homogeneity' in c and 'day2' in c]
    perim_day2 = [c for c in X.columns if 'Perimeter' in c and 'day2' in c]

    # Relative area growth from day0 to day2 (normalised to avoid scale sensitivity)
    if area_day0 and area_day2:
        X['area_growth_0to2'] = (X[area_day2[0]] - X[area_day0[0]]) / (X[area_day0[0]] + 1)
    # Relative area growth from day1 to day2 (captures late-stage dynamics)
    if area_day1 and area_day2:
        X['area_growth_1to2'] = (X[area_day2[0]] - X[area_day1[0]]) / (X[area_day1[0]] + 1)

    # Absolute and relative grey-value change (proxy for internal texture shift)
    if grey_day0 and grey_day2:
        X['grey_change_0to2'] = X[grey_day2[0]] - X[grey_day0[0]]
        X['grey_ratio_0to2']  = X[grey_day2[0]] / (X[grey_day0[0]] + 1)
    if grey_day1 and grey_day2:
        X['grey_change_1to2'] = X[grey_day2[0]] - X[grey_day1[0]]

    # Compactness at day2: 1 for a perfect circle, <1 for irregular shapes
    if area_day2 and perim_day2:
        X['compactness_day2'] = (4 * np.pi * X[area_day2[0]]) / (X[perim_day2[0]]**2 + 1)

    # Combined texture×size feature at day2
    if homog_day2 and area_day2:
        X['homog_area_day2'] = X[homog_day2[0]] * X[area_day2[0]]

    # Interaction term: does grey-value change correlate with area growth?
    if 'grey_change_0to2' in X.columns and 'area_growth_0to2' in X.columns:
        X['grey_area_interaction'] = X['grey_change_0to2'] * X['area_growth_0to2']

    return X

X_features = engineer_features(X_features)
print(f"  Train total features: {X_features.shape[1]}")

if CROSS_DATASET_MODE:
    X_features_test = engineer_features(X_features_test)
    print(f"  Test  total features: {X_features_test.shape[1]}")
    print(X_features_test)
```

## 4. Feature Selection (Top 30)

```python
print("\n[4/7] Feature selection...")

# If NAN_STRATEGY='drop' removed rows, realign df_complete to the surviving rows.
# Use .loc (label-based) not .iloc (position-based) – df_complete may have a
# non-contiguous integer index after filtering, so iloc would match wrong rows.
if len(X_features) < len(df_complete):
    df_complete = df_complete.loc[X_features.index].copy()
    X_features  = X_features.reset_index(drop=True)
    df_complete = df_complete.reset_index(drop=True)

# Binary target: 1 = Alive, 0 = Dead
# Strip whitespace in case of trailing spaces in the CSV label column
y = (df_complete['Label_day2'].str.strip() == 'Alive').astype(int)

print(f"  Samples: {len(y)}  |  Alive: {y.sum()}  |  Dead: {(y==0).sum()}")

# Diagnostic: show label breakdown per temperature – useful to check that
# control samples (e.g. 37 °C) are not accidentally labeled Dead
if MAX_TEMP_COL in df_complete.columns:
    print("\n  Label distribution by temperature:")
    temp_label = df_complete.groupby([MAX_TEMP_COL, df_complete['Label_day2'].str.strip()]).size().unstack(fill_value=0)
    print(temp_label.to_string())
    print()

if USE_FEATURE_SELECTION:
    # SelectKBest with mutual information ranks features by their statistical dependency
    # with the target label – keeps the top N_FEATURES most informative ones
    selector = SelectKBest(mutual_info_classif, k=N_FEATURES)
    X_selected = selector.fit_transform(X_features, y)
    selected_feature_names = X_features.columns[selector.get_support()].tolist()
    print(f"  Selected {X_selected.shape[1]} / {X_features.shape[1]} features (SelectKBest)")
    print(f"  Top 5: {selected_feature_names[:5]}")
else:
    # No selection – pass all engineered features directly to the model
    X_selected = X_features.values
    selected_feature_names = X_features.columns.tolist()
    selector = None
    print(f"  Feature selection disabled – using all {X_selected.shape[1]} features")

if CROSS_DATASET_MODE:
    if len(X_features_test) < len(df_complete_test):
        df_complete_test = df_complete_test.loc[X_features_test.index].copy()
        X_features_test  = X_features_test.reset_index(drop=True)
        df_complete_test = df_complete_test.reset_index(drop=True)
    y_test_ext = (df_complete_test['Label_day2'].str.strip() == 'Alive').astype(int)

    # Align test set to only the columns present in both train and test.
    # Features missing from the test set are dropped entirely (no imputation).
    train_cols = X_features.columns[selector.get_support()].tolist() if USE_FEATURE_SELECTION else X_features.columns.tolist()
    missing_cols = [col for col in train_cols if col not in X_features_test.columns]
    if missing_cols:
        print(f"  WARNING: {len(missing_cols)} feature(s) absent from test set and excluded: {missing_cols}")
    common_cols = [col for col in train_cols if col in X_features_test.columns]
    # Restrict training features to the same common columns so shapes match
    X_selected = X_features[common_cols].values
    selected_feature_names = common_cols
    X_selected_test = X_features_test[common_cols].values
    print(f"  Common features used (train & test): {len(common_cols)}")
else:
    X_selected_test = None
    y_test_ext      = None
```

## 5. K-Fold Cross-Validation (Random Forest)

```python
print("\n[5/7] Running 5-Fold Cross-Validation...")

n_splits = 5
kfold = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

fold_accuracies,       fold_precisions, fold_recalls  = [], [], []
fold_f1_scores,        fold_auc_scores                = [], []
fold_train_accuracies  = []   # for overfitting analysis

all_y_true, all_y_pred, all_y_proba, all_fold_indices = [], [], [], []
confusion_matrices   = []
fold_val_indices     = []
fold_val_predictions = []

rf_oof_pred  = np.full(len(df_complete), np.nan)
rf_oof_proba = np.full(len(df_complete), np.nan)

print("\n  Fold | Train Acc | Val Acc  | Gap     | Precision | Recall | F1-Score | AUC")
print("  " + "-"*85)

for fold_idx, (train_idx, val_idx) in enumerate(kfold.split(X_selected, y), 1):
    X_train_fold, X_val_fold = X_selected[train_idx], X_selected[val_idx]
    y_train_fold, y_val_fold = y.iloc[train_idx], y.iloc[val_idx]

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train_fold)
    X_val_scaled   = scaler.transform(X_val_fold)

    rf_model = RandomForestClassifier(
        n_estimators=300, max_depth=10, min_samples_split=5,
        class_weight='balanced', random_state=42, n_jobs=-1
    )
    rf_model.fit(X_train_scaled, y_train_fold)

    # Validation predictions
    y_pred_fold  = rf_model.predict(X_val_scaled)
    y_proba_fold = rf_model.predict_proba(X_val_scaled)[:, 1]

    # Train predictions (for overfitting check)
    y_pred_train_fold = rf_model.predict(X_train_scaled)
    train_acc = accuracy_score(y_train_fold, y_pred_train_fold)

    acc       = accuracy_score(y_val_fold, y_pred_fold)
    prec      = precision_score(y_val_fold, y_pred_fold)
    rec       = recall_score(y_val_fold, y_pred_fold)
    f1        = f1_score(y_val_fold, y_pred_fold)
    auc_score = roc_auc_score(y_val_fold, y_proba_fold)

    fold_train_accuracies.append(train_acc)
    fold_accuracies.append(acc);  fold_precisions.append(prec)
    fold_recalls.append(rec);     fold_f1_scores.append(f1)
    fold_auc_scores.append(auc_score)
    confusion_matrices.append(confusion_matrix(y_val_fold, y_pred_fold))

    all_y_true.extend(y_val_fold);   all_y_pred.extend(y_pred_fold)
    all_y_proba.extend(y_proba_fold); all_fold_indices.extend([fold_idx] * len(y_val_fold))

    fold_val_indices.append(val_idx)
    fold_val_predictions.append(y_pred_fold)

    rf_oof_pred[val_idx]  = y_pred_fold
    rf_oof_proba[val_idx] = y_proba_fold

    gap = train_acc - acc
    print(f"  {fold_idx:4d} | {train_acc:9.4f} | {acc:8.4f} | {gap:+.4f} | {prec:9.4f} | {rec:6.4f} | {f1:8.4f} | {auc_score:7.4f}")
```

## 6. Aggregate Cross-Validation Results

```python
print("\n[6/7] Aggregate Cross-Validation Results:")
print("  " + "-"*65)
print(f"  Mean Accuracy:  {np.mean(fold_accuracies):.4f} ± {np.std(fold_accuracies):.4f}")
print(f"  Mean Precision: {np.mean(fold_precisions):.4f} ± {np.std(fold_precisions):.4f}")
print(f"  Mean Recall:    {np.mean(fold_recalls):.4f} ± {np.std(fold_recalls):.4f}")
print(f"  Mean F1-Score:  {np.mean(fold_f1_scores):.4f} ± {np.std(fold_f1_scores):.4f}")
print(f"  Mean AUC:       {np.mean(fold_auc_scores):.4f} ± {np.std(fold_auc_scores):.4f}")

all_y_true  = np.array(all_y_true)
all_y_pred  = np.array(all_y_pred)
all_y_proba = np.array(all_y_proba)

overall_cm = confusion_matrix(all_y_true, all_y_pred)
print(f"\n  Overall Confusion Matrix (all folds combined):")
print(f"              Predicted")
print(f"              Dead  Alive")
print(f"  Actual Dead   {overall_cm[0,0]:3d}    {overall_cm[0,1]:3d}")
print(f"         Alive  {overall_cm[1,0]:3d}    {overall_cm[1,1]:3d}")
```

## 6b. Overfitting Analysis

```python
print("\n[6b] Overfitting Analysis — Random Forest")

OVERFIT_THRESHOLD = 0.10  # gap > 10% is flagged as overfitting

folds       = np.arange(1, n_splits + 1)
gaps        = [tr - val for tr, val in zip(fold_train_accuracies, fold_accuracies)]
mean_gap    = np.mean(gaps)
overfit_flag = mean_gap > OVERFIT_THRESHOLD

print(f"\n  {'Fold':>4} | {'Train Acc':>9} | {'Val Acc':>8} | {'Gap':>7}")
print("  " + "-"*40)
for i, (tr, val, g) in enumerate(zip(fold_train_accuracies, fold_accuracies, gaps), 1):
    flag = "  ← overfit?" if g > OVERFIT_THRESHOLD else ""
    print(f"  {i:4d} | {tr:9.4f} | {val:8.4f} | {g:+.4f}{flag}")
print(f"\n  Mean train acc : {np.mean(fold_train_accuracies):.4f}")
print(f"  Mean val acc   : {np.mean(fold_accuracies):.4f}")
print(f"  Mean gap       : {mean_gap:+.4f}  ({'⚠ OVERFITTING DETECTED' if overfit_flag else '✓ OK'})")

# ── Visualisation ────────────────────────────────────────────────────────────
fig_of, axes_of = plt.subplots(1, 2, figsize=(13, 4.5))

# Left: train vs val accuracy per fold
ax_l = axes_of[0]
ax_l.plot(folds, fold_train_accuracies, 'o-', color='steelblue',  linewidth=2, markersize=7, label='Train')
ax_l.plot(folds, fold_accuracies,       's-', color='coral',      linewidth=2, markersize=7, label='Validation')
ax_l.fill_between(folds, fold_train_accuracies, fold_accuracies,
                  alpha=0.15, color='red', label='Gap')
ax_l.axhline(np.mean(fold_train_accuracies), color='steelblue', linestyle='--', alpha=0.5)
ax_l.axhline(np.mean(fold_accuracies),       color='coral',     linestyle='--', alpha=0.5)
ax_l.set_xlabel('Fold', fontsize=11); ax_l.set_ylabel('Accuracy', fontsize=11)
ax_l.set_title('RF — Train vs Validation Accuracy per Fold', fontsize=12, fontweight='bold')
ax_l.set_xticks(folds); ax_l.set_ylim([0.5, 1.05])
ax_l.legend(); ax_l.grid(True, alpha=0.3)

# Right: gap per fold bar chart
ax_r = axes_of[1]
colors_bar = ['#d62728' if g > OVERFIT_THRESHOLD else '#2ca02c' for g in gaps]
ax_r.bar(folds, gaps, color=colors_bar, alpha=0.8, edgecolor='black', linewidth=0.8)
ax_r.axhline(OVERFIT_THRESHOLD, color='red', linestyle='--', linewidth=1.5,
             label=f'Threshold ({OVERFIT_THRESHOLD:.0%})')
ax_r.axhline(0, color='black', linewidth=0.8)
ax_r.set_xlabel('Fold', fontsize=11); ax_r.set_ylabel('Train − Val Accuracy', fontsize=11)
ax_r.set_title('RF — Overfitting Gap per Fold', fontsize=12, fontweight='bold')
ax_r.set_xticks(folds); ax_r.legend(); ax_r.grid(True, alpha=0.3, axis='y')
for i, g in enumerate(gaps):
    ax_r.text(i + 1, g + 0.003, f'{g:+.3f}', ha='center', fontsize=9)

status_color = '#d62728' if overfit_flag else '#2ca02c'
fig_of.suptitle(
    f'RF Overfitting Analysis — Mean gap: {mean_gap:+.4f}  '
    f'{"⚠ Overfitting detected" if overfit_flag else "✓ No significant overfitting"}',
    fontsize=13, fontweight='bold', color=status_color
)
plt.tight_layout()
plt.show()
print("  ✓ Overfitting analysis complete")
```

## 8a. Temperature vs Predicted Label — Helper Function & Best-Fold Plot

```python
import numpy as np
import scipy.optimize
import emcee
import matplotlib.pyplot as plt

def sigmoid(x, IC50, slope):
    with np.errstate(divide='ignore', invalid='ignore'):
        return 1 / (1 + (IC50 / x)**slope)

def log_likelihood(params, x, y):
    IC50, slope = params
    y_pred = sigmoid(x, IC50, slope)
    return -0.5 * np.sum((y - y_pred)**2)

def log_prior(params):
    IC50, slope = params
    if 30 < IC50 < 60 and -50 < slope < -0.1:
        return 0.0
    return -np.inf

def log_posterior(params, x, y):
    lp = log_prior(params)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_likelihood(params, x, y)

def IC50_pred(X, Y):
    nwalkers = 32
    ndim = 2
    pos = np.random.uniform([30, -50], [60, -0.1], (nwalkers, ndim))
    sampler = emcee.EnsembleSampler(nwalkers, ndim, log_posterior, args=(X, Y))
    sampler.run_mcmc(pos, 5000)
    samples = sampler.chain[:, 100:, :].reshape((-1, ndim))
    IC50_mcmc = np.percentile(samples[:, 0], [16, 50, 68])
    slope_mcmc = np.percentile(samples[:, 1], [16, 50, 68])
    return IC50_mcmc, slope_mcmc

def getConcentrationPrediction(test_index_list, y_pred_list, df, col='Concentration (uM)'):
    concentration_list = []
    prediction_list = []
    for test_index, y_pred in zip(test_index_list, y_pred_list):
        concentration = df.loc[test_index, col].values
        prediction = y_pred
        concentration_list.extend(concentration)
        prediction_list.extend(prediction)
    return np.array(concentration_list), np.array(prediction_list)

def plot_dose_response_IC50(ax, test_index_list, y_pred_list, df, kws_dict, group, col='Concentration (uM)', unit='μM', ylabel='Predicted Viability'):
    X, Y = getConcentrationPrediction(test_index_list, y_pred_list, df, col)
    IC50_mcmc, slope_mcmc = IC50_pred(X, Y)
    x_range = np.linspace(min(X), max(X), 100)
    y_fit = sigmoid(x_range, IC50_mcmc[1], slope_mcmc[1])
    ax.scatter(X, Y, **kws_dict)
    ax.plot(x_range, y_fit, 'r-', linewidth=2, label=f'IC50 = {IC50_mcmc[1]:.2f} {unit}\nSlope = {slope_mcmc[1]:.2f}')
    ax.set_xlabel(f'{col} ({unit})')
    ax.set_ylabel(ylabel)
    ax.set_title(f'Dose-Response Curve ({group})')
    ax.legend()
    ax.grid(True, alpha=0.3)
    return IC50_mcmc, slope_mcmc

def run_bayesian_ic50(X, Y):
    """
    Standardized MCMC runner to match 3_machine_learning_spheroids_morpho_MCF7.ipynb
    """
    X = np.array([0.01 if x <= 0.001 else x for x in X])
    Y = np.array(Y)

    # Setup walkers
    nwalkers = 32
    ndim = 2
    
    # Initialize walkers in a broad range matching the prior
    p0 = np.random.rand(nwalkers, ndim)
    p0[:, 0] = p0[:, 0] * 20 + 30  # IC50 guess around 30-50
    p0[:, 1] = p0[:, 1] * 10 - 15  # Slope guess around -15 to -5

    sampler = emcee.EnsembleSampler(nwalkers, ndim, log_posterior, args=(X, Y))
    
    # Run with 5000 steps as requested
    sampler.run_mcmc(p0, 5000, progress=False) 

    # Extract samples (discard 1000 for burn-in for better stability)
    samples = sampler.get_chain(discard=1000, thin=15, flat=True)
    
    # Median estimates
    IC50_mcmc, slope_mcmc = np.percentile(samples, 50, axis=0)
    return IC50_mcmc, slope_mcmc, samples
```

```python
def plot_temp_vs_predicted(temp_values, pred_labels, title, ax=None, color='steelblue'):
    """Predicted label vs Max Temperature using Bayesian MCMC for IC50."""
    from matplotlib.lines import Line2D
    
    standalone = ax is None
    if standalone:
        fig_out, ax = plt.subplots(figsize=(10, 6))
    else:
        fig_out = ax.get_figure()

    temp_arr = np.array(temp_values, dtype=float)
    pred_arr = np.array(pred_labels)
    mask = ~np.isnan(temp_arr)
    temp_clean, pred_clean = temp_arr[mask], pred_arr[mask]

    if len(np.unique(pred_clean)) < 2:
        ax.set_title(title + " (Insufficient data)")
        return fig_out, ax, None

    # --- Run the Bayesian MCMC ---
    ic50, slope, samples = run_bayesian_ic50(temp_clean, pred_clean)

    # Plotting the fit
    t_range = np.linspace(temp_clean.min()-2, temp_clean.max()+2, 200)
    p_fit = sigmoid(t_range, ic50, slope)
    ax.plot(t_range, p_fit, color=color, linewidth=3, label=f'Bayesian MCMC Fit')

    # Visualizing Uncertainty (optional but recommended for MCMC)
    for i in np.random.randint(len(samples), size=100):
        sample_ic50, sample_slope = samples[i]
        ax.plot(t_range, sigmoid(t_range, sample_ic50, sample_slope), color=color, alpha=0.05)

    # Scatter and styling
    rng = np.random.default_rng(42)
    ax.scatter(temp_clean + rng.uniform(-0.2, 0.2, len(temp_clean)), 
               pred_clean + rng.uniform(-0.03, 0.03, len(pred_clean)), 
               c=['#e74c3c' if p == 0 else '#2ecc71' for p in pred_clean], 
               alpha=0.6, edgecolors='k', zorder=3)

    if ic50:
        ax.axvline(ic50, color='darkorange', linestyle='--', label=f'IC50 = {ic50:.2f} °C')
    
    ax.set_xlabel('Temperature (°C)'); ax.set_ylabel('Probability Alive')
    ax.set_title(title)
    ax.legend()
    return fig_out, ax, ic50
```

## 8b. Neural Network — Training & Evaluation

```python
print("\n[8b/8] Training Neural Network...")
from tensorflow.keras.regularizers import l2 as keras_l2

num_classes = 2
size = X_train_final_scaled.shape[1]

# One-hot encode labels for categorical cross-entropy (e.g. 1 → [0,1], 0 → [1,0])
y_train_nn = tf.keras.utils.to_categorical(y_train_final, num_classes)
y_test_nn  = tf.keras.utils.to_categorical(y_test_final,  num_classes)

# L2 regularization penalises large weights — helps generalisation
_reg = keras_l2(L2_REG) if L2_REG > 0 else None

# Build model — Dropout layers are inserted only when DROPOUT_RATE > 0
model = tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(size,)),
    tf.keras.layers.Dense(128, activation='relu', kernel_regularizer=_reg),
    *([ tf.keras.layers.Dropout(DROPOUT_RATE) ] if DROPOUT_RATE > 0 else []),
    tf.keras.layers.Dense(10,  activation='relu', kernel_regularizer=_reg),
    *([ tf.keras.layers.Dropout(DROPOUT_RATE) ] if DROPOUT_RATE > 0 else []),
    tf.keras.layers.Dense(num_classes, activation='softmax')  # output: P(Dead), P(Alive)
])
model.compile(optimizer='Adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Early stopping monitors val_loss and restores the best weights when it stops improving
callbacks = []
if EARLY_STOPPING:
    callbacks.append(tf.keras.callbacks.EarlyStopping(
        monitor='val_loss', patience=EARLY_STOPPING_PATIENCE,
        restore_best_weights=True, verbose=1
    ))

# When early stopping is on, allow up to 100 epochs; otherwise fix at 30
history = model.fit(
    X_train_final_scaled, y_train_nn,
    epochs=100 if EARLY_STOPPING else 30,
    batch_size=32, validation_split=0.1,   # 10% of training data used for validation
    callbacks=callbacks, verbose=0
)

nn_proba       = model.predict(X_test_final_scaled, verbose=0)
nn_pred        = np.argmax(nn_proba, axis=1)       # class with highest probability
nn_proba_alive = nn_proba[:, 1]                    # P(Alive) for ROC/AUC

nn_acc  = accuracy_score(y_test_final, nn_pred)
nn_prec = precision_score(y_test_final, nn_pred)
nn_rec  = recall_score(y_test_final, nn_pred)
nn_f1   = f1_score(y_test_final, nn_pred)
nn_auc  = roc_auc_score(y_test_final, nn_proba_alive)

epochs_run = len(history.history['loss'])
print(f"\n  Epochs run     : {epochs_run} {'(early stopped)' if EARLY_STOPPING and epochs_run < 100 else ''}")
print(f"  Dropout        : {DROPOUT_RATE}")
print(f"  L2 reg         : {L2_REG}")
print(f"\n  NEURAL NETWORK TEST PERFORMANCE:")
print(f"  Accuracy:  {nn_acc:.4f}")
print(f"  Precision: {nn_prec:.4f}")
print(f"  Recall:    {nn_rec:.4f}")
print(f"  F1-Score:  {nn_f1:.4f}")
print(f"  ROC AUC:   {nn_auc:.4f}")
print(f"\n  Winner: {'Random Forest' if final_acc > nn_acc else 'Neural Network' if nn_acc > final_acc else 'Tie'} (Accuracy)")
```

## 8b-ii. Neural Network — 5-Fold Cross-Validation

```python
print("\n[8b-ii] NN 5-Fold Cross-Validation (same splits as RF)...")

nn_fold_accuracies,      nn_fold_precisions, nn_fold_recalls = [], [], []
nn_fold_f1_scores,       nn_fold_auc_scores                  = [], []
nn_fold_train_accuracies = []   # for overfitting analysis

nn_oof_pred  = np.full(len(df_complete), np.nan)
nn_oof_proba = np.full(len(df_complete), np.nan)

for fold_idx, (train_idx, val_idx) in enumerate(kfold.split(X_selected, y), 1):
    X_tr, X_vl = X_selected[train_idx], X_selected[val_idx]
    y_tr, y_vl = y.iloc[train_idx],    y.iloc[val_idx]

    sc = StandardScaler()
    X_tr_s = sc.fit_transform(X_tr)
    X_vl_s = sc.transform(X_vl)

    _reg_fold = keras_l2(L2_REG) if L2_REG > 0 else None

    nn_fold = tf.keras.models.Sequential([
        tf.keras.layers.Input(shape=(X_tr_s.shape[1],)),
        tf.keras.layers.Dense(128, activation='relu', kernel_regularizer=_reg_fold),
        *([ tf.keras.layers.Dropout(DROPOUT_RATE) ] if DROPOUT_RATE > 0 else []),
        tf.keras.layers.Dense(10,  activation='relu', kernel_regularizer=_reg_fold),
        *([ tf.keras.layers.Dropout(DROPOUT_RATE) ] if DROPOUT_RATE > 0 else []),
        tf.keras.layers.Dense(2,   activation='softmax'),
    ])
    nn_fold.compile(optimizer='Adam', loss='categorical_crossentropy', metrics=['accuracy'])

    fold_callbacks = []
    if EARLY_STOPPING:
        fold_callbacks.append(tf.keras.callbacks.EarlyStopping(
            monitor='val_loss', patience=EARLY_STOPPING_PATIENCE,
            restore_best_weights=True, verbose=0
        ))

    nn_fold.fit(
        X_tr_s, tf.keras.utils.to_categorical(y_tr, 2),
        epochs=100 if EARLY_STOPPING else 30,
        batch_size=32, validation_split=0.1,
        callbacks=fold_callbacks, verbose=0
    )

    nn_proba_fold = nn_fold.predict(X_vl_s, verbose=0)
    nn_pred_fold  = np.argmax(nn_proba_fold, axis=1)
    nn_prob_alive = nn_proba_fold[:, 1]

    # Train accuracy for overfitting check
    nn_pred_train_fold = np.argmax(nn_fold.predict(X_tr_s, verbose=0), axis=1)
    nn_fold_train_accuracies.append(accuracy_score(y_tr, nn_pred_train_fold))

    nn_fold_accuracies.append(accuracy_score(y_vl, nn_pred_fold))
    nn_fold_precisions.append(precision_score(y_vl, nn_pred_fold, zero_division=0))
    nn_fold_recalls.append(recall_score(y_vl, nn_pred_fold, zero_division=0))
    nn_fold_f1_scores.append(f1_score(y_vl, nn_pred_fold, zero_division=0))
    nn_fold_auc_scores.append(roc_auc_score(y_vl, nn_prob_alive))

    nn_oof_pred[val_idx]  = nn_pred_fold
    nn_oof_proba[val_idx] = nn_prob_alive

    gap_nn = nn_fold_train_accuracies[-1] - nn_fold_accuracies[-1]
    print(f"    Fold {fold_idx}: TrainAcc={nn_fold_train_accuracies[-1]:.4f}  ValAcc={nn_fold_accuracies[-1]:.4f}  "
          f"Gap={gap_nn:+.4f}  F1={nn_fold_f1_scores[-1]:.4f}  AUC={nn_fold_auc_scores[-1]:.4f}")

print(f"  NN Mean Accuracy: {np.mean(nn_fold_accuracies):.4f} ± {np.std(nn_fold_accuracies):.4f}")
```

```python
print("\n[12] K-fold box plots (RF only, NN only, RF vs NN)...")

METRIC_LABELS  = ['Accuracy', 'Precision', 'Recall', 'F1-Score', 'AUC']
RF_METRIC_DATA = [fold_accuracies,    fold_precisions,    fold_recalls,
                  fold_f1_scores,     fold_auc_scores]
NN_METRIC_DATA = [nn_fold_accuracies, nn_fold_precisions, nn_fold_recalls,
                  nn_fold_f1_scores,  nn_fold_auc_scores]
RF_COLOR = '#2166ac'; NN_COLOR = '#d6604d'


def _paper_boxplot(ax, metric_data, metric_labels, color, model_name):
    bp = ax.boxplot(
        metric_data, positions=range(len(metric_labels)), widths=0.42,
        patch_artist=True, notch=False,
        medianprops=dict(color='black',   linewidth=2.2),
        whiskerprops=dict(color='#444444', linewidth=1.4, linestyle='--'),
        capprops=dict(color='#444444',    linewidth=1.8),
        flierprops=dict(marker='o', markersize=5, linestyle='none',
                        markeredgecolor='#666666', markerfacecolor='white', markeredgewidth=1.0),
        boxprops=dict(linewidth=1.4),
    )
    for patch in bp['boxes']:
        patch.set_facecolor(color); patch.set_alpha(0.65)

    rng = np.random.default_rng(7)
    for i, vals in enumerate(metric_data):
        jit = rng.uniform(-0.11, 0.11, size=len(vals))
        ax.scatter(np.full(len(vals), i) + jit, vals, color=color, s=52, zorder=5,
                   edgecolors='black', linewidths=0.7, alpha=0.88)

    for i, vals in enumerate(metric_data):
        m, s = np.mean(vals), np.std(vals)
        ax.text(i, max(vals) + 0.016, f'{m:.3f}\n±{s:.3f}', ha='center', va='bottom',
                fontsize=8.5, color='#111111', fontweight='bold', linespacing=1.3)

    ax.set_xticks(range(len(metric_labels))); ax.set_xticklabels(metric_labels, fontsize=12)
    ax.set_ylabel('Score', fontsize=13, labelpad=8)
    ax.set_title(f'{n_splits}-Fold CV — {model_name}', fontsize=14, fontweight='bold', pad=12)

    all_flat = [v for vals in metric_data for v in vals]
    ax.set_ylim(0, min(1.01, max(all_flat) + 0.13))
    ax.yaxis.set_major_locator(mticker.MultipleLocator(0.05))
    ax.tick_params(axis='both', labelsize=11, direction='out', length=5)
    for sp in ['top', 'right']: ax.spines[sp].set_visible(False)
    for sp in ['left', 'bottom']: ax.spines[sp].set_linewidth(1.4)
    ax.grid(axis='y', linestyle='--', linewidth=0.7, alpha=0.45, zorder=0)


# RF only
fig_rf_bp, ax_rf_bp = plt.subplots(figsize=(8.5, 6.0))
_paper_boxplot(ax_rf_bp, RF_METRIC_DATA, METRIC_LABELS, RF_COLOR, 'Random Forest')
plt.tight_layout()
rf_bp_path = OUTPUT_DIR / 'kfold_metrics_boxplot_RF.png'
plt.savefig(rf_bp_path, dpi=300, bbox_inches='tight', facecolor='white')
plt.show(); plt.close()
print(f"  ✓ RF box plot saved → {rf_bp_path}")

# NN only
fig_nn_bp, ax_nn_bp = plt.subplots(figsize=(8.5, 6.0))
_paper_boxplot(ax_nn_bp, NN_METRIC_DATA, METRIC_LABELS, NN_COLOR, 'Neural Network')
plt.tight_layout()
nn_bp_path = OUTPUT_DIR / 'kfold_metrics_boxplot_NN.png'
plt.savefig(nn_bp_path, dpi=300, bbox_inches='tight', facecolor='white')
plt.show(); plt.close()
print(f"  ✓ NN box plot saved → {nn_bp_path}")

# RF vs NN side by side
from matplotlib.patches import Patch

n_metrics  = len(METRIC_LABELS)
GROUP_GAP  = 1.1
BOX_OFFSET = 0.24
rf_positions = [i * GROUP_GAP - BOX_OFFSET for i in range(n_metrics)]
nn_positions = [i * GROUP_GAP + BOX_OFFSET for i in range(n_metrics)]

fig_cmp, ax_cmp = plt.subplots(figsize=(11.5, 6.5))

def _side_boxplot(ax, positions, metric_data, color, label):
    bp = ax.boxplot(
        metric_data, positions=positions, widths=0.38, patch_artist=True, notch=False,
        medianprops=dict(color='black',   linewidth=2.2),
        whiskerprops=dict(color='#444444', linewidth=1.4, linestyle='--'),
        capprops=dict(color='#444444',    linewidth=1.8),
        flierprops=dict(marker='o', markersize=5, linestyle='none',
                        markeredgecolor='#666666', markerfacecolor='white', markeredgewidth=1.0),
        boxprops=dict(linewidth=1.4),
    )
    for patch in bp['boxes']:
        patch.set_facecolor(color); patch.set_alpha(0.65)

    rng = np.random.default_rng(13)
    for pos, vals in zip(positions, metric_data):
        jit = rng.uniform(-0.07, 0.07, size=len(vals))
        ax.scatter(np.full(len(vals), pos) + jit, vals, color=color, s=48, zorder=5,
                   edgecolors='black', linewidths=0.7, alpha=0.88)

    for pos, vals in zip(positions, metric_data):
        m, s = np.mean(vals), np.std(vals)
        ax.text(pos, max(vals) + 0.016, f'{m:.3f}\n±{s:.3f}', ha='center', va='bottom',
                fontsize=7.8, color='#111111', fontweight='bold', linespacing=1.3)

    return Patch(facecolor=color, alpha=0.65, edgecolor='black', linewidth=1.2, label=label)

proxy_rf = _side_boxplot(ax_cmp, rf_positions, RF_METRIC_DATA, RF_COLOR, 'Random Forest')
proxy_nn = _side_boxplot(ax_cmp, nn_positions, NN_METRIC_DATA, NN_COLOR, 'Neural Network')

for i in range(n_metrics):
    if i % 2 == 0:
        ax_cmp.axvspan(i*GROUP_GAP - GROUP_GAP/2, i*GROUP_GAP + GROUP_GAP/2,
                       color='#f0f0f0', zorder=0, linewidth=0)

group_centers = [i * GROUP_GAP for i in range(n_metrics)]
ax_cmp.set_xticks(group_centers); ax_cmp.set_xticklabels(METRIC_LABELS, fontsize=12.5)
ax_cmp.set_xlim(-GROUP_GAP*0.55, (n_metrics-1)*GROUP_GAP + GROUP_GAP*0.55)
ax_cmp.set_ylabel('Score', fontsize=13, labelpad=8)
ax_cmp.set_title(f'{n_splits}-Fold CV — Random Forest vs Neural Network',
                 fontsize=15, fontweight='bold', pad=14)
all_flat_cmp = [v for md in [RF_METRIC_DATA, NN_METRIC_DATA] for vals in md for v in vals]
ax_cmp.set_ylim(0, min(1.0, max(all_flat_cmp) + 0.14))
ax_cmp.yaxis.set_major_locator(mticker.MultipleLocator(0.05))
ax_cmp.tick_params(axis='both', labelsize=11, direction='out', length=5)
for sp in ['top', 'right']: ax_cmp.spines[sp].set_visible(False)
for sp in ['left', 'bottom']: ax_cmp.spines[sp].set_linewidth(1.4)
ax_cmp.grid(axis='y', linestyle='--', linewidth=0.7, alpha=0.45, zorder=0)
ax_cmp.legend(handles=[proxy_rf, proxy_nn], fontsize=11, frameon=True, loc='lower right',
              framealpha=0.92, edgecolor='#cccccc', borderpad=0.9, handlelength=1.6)

plt.tight_layout()
cmp_bp_path = OUTPUT_DIR / 'kfold_metrics_boxplot_RF_vs_NN.png'
plt.savefig(cmp_bp_path, dpi=300, bbox_inches='tight', facecolor='white')
plt.show(); plt.close()
print(f"  ✓ RF vs NN box plot saved → {cmp_bp_path}")

plt.rcParams.update(plt.rcParamsDefault)
```

```python
# ═══════════════════════════════════════════════════════════════════════════════
# SECTION: Multi-Cell-Line Comparison  (CROSS_DATASET_MODE = False)
#          Interactive Plots  (CROSS_DATASET_MODE = True)
# ═══════════════════════════════════════════════════════════════════════════════
# HOW TO USE (CROSS_DATASET_MODE = True):
#   1. Ensure all prior cells have been run.
#   2. Pick a color and adjust transparency, then click Update Plots.
# HOW TO USE (CROSS_DATASET_MODE = False):
#   1. Fill in CELL_LINE_CSVS in the config cell.
#   2. Run all cells top-to-bottom, then run this cell.
# ═══════════════════════════════════════════════════════════════════════════════

# ── FONT SIZE CONFIGURATION ──────────────────────────────────────────────────────
# Adjust these values to resize all text in the plots at once.
_FS_SUPTITLE      = 10   # main figure title
_FS_AXES_TITLE    = 18   # individual subplot titles
_FS_LABEL         = 18   # x / y axis labels
_FS_TICK          = 18   # tick labels (both axes)
_FS_LEGEND        = 16   # legend text
_FS_LEGEND_TITLE  = 16   # legend title
_FS_BOXPLOT_STAT  = 10   # mean ± std text shown above each box
_FS_GAP_TEXT      = 12    # train / validation gap annotation
# ─────────────────────────────────────────────────────────────────────────────

from matplotlib import artist


if CROSS_DATASET_MODE:
    import ipywidgets as widgets
    import matplotlib
    from IPython.display import display as _ipy_display

    _dataset_name = f"{Path(TRAIN_CSV).stem} → {Path(TEST_CSV).stem}"
    _save_dir     = OUTPUT_DIR
    _rng          = np.random.default_rng(42)

    _METRIC_LABELS = ['Accuracy', 'Precision', 'Recall', 'F1', 'AUC']
    _RF_DATA = [fold_accuracies, fold_precisions, fold_recalls, fold_f1_scores, fold_auc_scores]
    _NN_DATA = [nn_fold_accuracies, nn_fold_precisions, nn_fold_recalls,
                nn_fold_f1_scores, nn_fold_auc_scores]

    # ── Times New Roman font settings ──────────────────────────────────────────
    _tnr_rc = {
        'font.family':           'serif',
        'font.serif':            ['Times New Roman'],
        'axes.titlesize':        _FS_AXES_TITLE,
        'axes.titleweight':      'bold',
        'axes.labelsize':        _FS_LABEL,
        'axes.labelweight':      'bold',
        'xtick.labelsize':       _FS_TICK,
        'ytick.labelsize':       _FS_TICK,
        'legend.fontsize':       _FS_LEGEND,
        'legend.title_fontsize': _FS_LEGEND_TITLE,
    }

    # ── Widgets ───────────────────────────────────────────────────────────────────────────
    _color_picker = widgets.ColorPicker(
        description=_dataset_name,
        value='#2166ac',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='420px')
    )
    _dead_alpha_slider = widgets.FloatSlider(
        value=0.55, min=0.0, max=1.0, step=0.05,
        description='Dead α:',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='380px')
    )
    _alive_alpha_slider = widgets.FloatSlider(
        value=0.25, min=0.0, max=1.0, step=0.05,
        description='Alive α:',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='380px')
    )
    _run_btn = widgets.Button(
        description='▶  Update Plots',
        button_style='primary',
        layout=widgets.Layout(width='180px', margin='10px 0 0 0')
    )
    _out = widgets.Output()
    _ipy_display(widgets.VBox([
        widgets.HTML(
            '<b style="font-size:14px; font-family: Times New Roman">'
            'Select a color and adjust violin transparency, then click Update Plots:</b>'
        ),
        _color_picker,
        widgets.HTML('<i style="font-size:12px">Violin transparency (violin plot only):</i>'),
        _dead_alpha_slider,
        _alive_alpha_slider,
        _run_btn,
        _out,
    ]))

    def _do_plots_cross(b=None):
        with _out:
            _out.clear_output(wait=True)

        _col        = matplotlib.colors.to_rgba(_color_picker.value)
        _col3       = tuple(_col[:3])
        _dead_alpha = _dead_alpha_slider.value
        _alive_alpha = _alive_alpha_slider.value

        with matplotlib.rc_context(_tnr_rc):

            # ── Figures 1a & 1b: RF and NN k-fold validation metrics ────────────────
            for _model_name, _metric_data in [('Random Forest', _RF_DATA),
                                               ('Neural Network', _NN_DATA)]:
                fig, ax = plt.subplots(figsize=(8.5, 6.0))
                fig.suptitle(
                    f"{n_splits}-Fold Validation Metrics — {_model_name}\n[{_dataset_name}]",
                    fontsize=_FS_SUPTITLE, fontweight='bold')
                bp = ax.boxplot(
                    _metric_data, positions=range(len(_METRIC_LABELS)), widths=0.42,
                    patch_artist=True, notch=False,
                    medianprops=dict(color='black', linewidth=2.2),
                    whiskerprops=dict(color='#444444', linewidth=1.4, linestyle='--'),
                    capprops=dict(color='#444444', linewidth=1.8),
                    flierprops=dict(marker='o', markersize=5, linestyle='none',
                                    markeredgecolor='#666666', markerfacecolor='white',
                                    markeredgewidth=1.0),
                    boxprops=dict(linewidth=1.4),
                )
                for patch in bp['boxes']:
                    patch.set_facecolor(_col)
                    patch.set_alpha(0.65)
                for i, vals in enumerate(_metric_data):
                    jit = _rng.uniform(-0.11, 0.11, size=len(vals))
                    ax.scatter(np.full(len(vals), i) + jit, vals, color=_col, s=52, zorder=5,
                               edgecolors='black', linewidths=0.7, alpha=0.88)
                ax.set_xticks(range(len(_METRIC_LABELS)))
                ax.set_xticklabels(_METRIC_LABELS, fontsize=_FS_TICK, fontweight='bold')
                ax.set_ylabel('Score', fontsize=_FS_LABEL, labelpad=8)
                _all_flat = [v for vals in _metric_data for v in vals]
                ax.set_ylim(0, min(1.01, max(_all_flat) + 0.13))
                ax.yaxis.set_major_locator(mticker.MultipleLocator(0.2))
                ax.tick_params(axis='both', labelsize=_FS_TICK, direction='out', length=5)
                for sp in ['top', 'right']:   ax.spines[sp].set_visible(False)
                for sp in ['left', 'bottom']: ax.spines[sp].set_linewidth(1.4)
                ax.grid(axis='y', linestyle='--', linewidth=0.7, alpha=0.45, zorder=0)
                plt.tight_layout()
                _fname = ('cross_kfold_RF_metrics.png' if 'Forest' in _model_name
                          else 'cross_kfold_NN_metrics.png')
                fig.savefig(_save_dir / _fname, dpi=300, bbox_inches='tight', facecolor='white')
                plt.show()
                print(f'  ✓ {_model_name} k-fold metrics saved → {_save_dir / _fname}')

            # ── Figure 2: Overfitting analysis — RF and NN side by side ───────────
            fig2, axes2 = plt.subplots(1, 2, figsize=(13, 5.5), sharey=True)
            fig2.suptitle(
                f"Overfitting Analysis: Train vs Validation Accuracy per Fold\n[{_dataset_name}]",
                fontsize=_FS_SUPTITLE, fontweight='bold')
            _tr_offset  = -0.18
            _val_offset =  0.18

            for _mi, (_model_label, _tr_accs, _val_accs) in enumerate([
                ('Random Forest',  fold_train_accuracies,    fold_accuracies),
                ('Neural Network', nn_fold_train_accuracies, nn_fold_accuracies),
            ]):
                ax = axes2[_mi]
                for _tv, _vv in zip(_tr_accs, _val_accs):
                    ax.plot([_tr_offset, _val_offset], [_tv, _vv],
                            color=_col3, alpha=0.35, linewidth=0.9, zorder=3)
                for _vals, _off, _alpha, _lw in [
                    (_tr_accs,  _tr_offset,  0.55, 2.0),
                    (_val_accs, _val_offset, 0.25, 1.5),
                ]:
                    ax.boxplot([_vals], positions=[_off], widths=0.22,
                               patch_artist=True,
                               boxprops=dict(facecolor=(*_col3, _alpha), color=_col3),
                               medianprops=dict(color=_col3, linewidth=_lw),
                               whiskerprops=dict(color=_col3, linewidth=1.0),
                               capprops=dict(color=_col3, linewidth=1.0),
                               flierprops=dict(marker='', markersize=0))
                    _jit = _rng.uniform(-0.06, 0.06, len(_vals))
                    ax.scatter(np.full(len(_vals), _off) + _jit, _vals, color=_col,
                               s=30, zorder=5, alpha=0.9, edgecolors='white', linewidths=0.4)
                _gap = np.mean(_tr_accs) - np.mean(_val_accs)
                ax.text(0, 0.02, f'mean gap\n{_gap:+.3f}', ha='center', va='bottom',
                        fontsize=_FS_GAP_TEXT, color='#555555', style='italic',
                        fontfamily='Times New Roman')
                ax.set_xticks([_tr_offset, _val_offset])
                ax.set_xticklabels(['Train', 'Validation'], fontsize=_FS_TICK)
                ax.set_xlim(-0.55, 0.55)
                ax.set_ylim(0, 1.08)
                ax.set_ylabel('Accuracy', fontsize=_FS_LABEL, fontweight='bold')
                ax.set_title(f'{_model_label} — Train vs Validation',
                             fontsize=_FS_AXES_TITLE, fontweight='bold')
                ax.axhline(0.5, color='gray', linestyle='--', linewidth=0.8, alpha=0.5)
                ax.grid(axis='y', alpha=0.3)
                ax.spines[['top', 'right']].set_visible(False)
                ax.tick_params(axis='both', labelsize=_FS_TICK)
                from matplotlib.patches import Patch
                ax.legend(
                    handles=[Patch(facecolor=(*_col3, 0.6), label='Train'),
                             Patch(facecolor=(*_col3, 0.2), label='Validation')],
                    fontsize=_FS_LEGEND, loc='lower right')

            plt.tight_layout()
            _oa_path = _save_dir / 'cross_overfitting_analysis.png'
            fig2.savefig(_oa_path, dpi=300, bbox_inches='tight', facecolor='white')
            plt.show()
            print(f'  ✓ Overfitting analysis saved → {_oa_path}')

            # ── Figure 3: Violin — temperature by NN predicted vs true label ──
            _temp_src = _df_test
            if MAX_TEMP_COL not in _temp_src.columns:
                print(f"  WARNING: '{MAX_TEMP_COL}' not found in test data → skipping violin plot.")
            else:
                _temps_v = _temp_src.iloc[_test_pos][MAX_TEMP_COL].values.astype(float)
                _true_v  = pd.Series(y_test_final).map({1: 'Alive', 0: 'Dead'})
                _nn_v    = pd.Series(nn_pred).map({1: 'Alive', 0: 'Dead'})

                _source_order = ['True label', 'NN']

                _rows_v = []
                for _tl, _nl, _t in zip(_true_v, _nn_v, _temps_v):
                    if not np.isnan(_t):
                        _rows_v.append({'Source': 'True label', 'Class': _tl, 'Temperature': _t})
                        _rows_v.append({'Source': 'NN',          'Class': _nl, 'Temperature': _t})
                _df_v = pd.DataFrame(_rows_v)

                fig_v, ax_v = plt.subplots(figsize=(9, 7))
                fig_v.suptitle(
                    f'Temperature Distribution — NN Predicted vs True Label\n[{_dataset_name}]',
                    fontsize=_FS_SUPTITLE, fontweight='bold')

                sns.violinplot(
                    data=_df_v[_df_v['Class'] == 'Dead'],
                    x='Source', y='Temperature', ax=ax_v,
                    order=_source_order, color='#d62728',
                    inner=None, linewidth=3, dodge=False, width=0.8
                )
                for artist in ax_v.collections:
                            artist.set_alpha(_dead_alpha)
                            artist.set_linestyle('--')
                sns.violinplot(
                    data=_df_v[_df_v['Class'] == 'Alive'],
                    x='Source', y='Temperature', ax=ax_v,
                    order=_source_order, color='#2ca02c',
                    inner=None, linewidth=3, dodge=False, width=0.8
                )
                for artist in ax_v.collections:
                            artist.set_alpha(_alive_alpha)

                sns.stripplot(
                    data=_df_v[_df_v['Class'] == 'Dead'],
                    x='Source', y='Temperature', ax=ax_v,
                    order=_source_order, dodge=True, jitter=True,
                    alpha=0.85, size=7, marker='^', color='#d62728',
                    linewidth=0.3, edgecolor='white', legend=False
                )
                sns.stripplot(
                    data=_df_v[_df_v['Class'] == 'Alive'],
                    x='Source', y='Temperature', ax=ax_v,
                    order=_source_order, dodge=True, jitter=True,
                    alpha=0.4, size=7, marker='o', color='#2ca02c',
                    linewidth=0.3, edgecolor='white', legend=False
                )

                ax_v.set_xlabel('Prediction Source', fontsize=_FS_LABEL, fontweight='bold')
                ax_v.set_ylabel('Temperature (°C)',   fontsize=_FS_LABEL, fontweight='bold')
                ax_v.set_ylim(20, 100)
                ax_v.grid(True, alpha=0.3, axis='y')
                ax_v.tick_params(axis='both', labelsize=_FS_TICK)
                ax_v.spines[['top', 'right']].set_visible(False)
                if ax_v.get_legend() is not None:
                    ax_v.get_legend().remove()

                from matplotlib.lines import Line2D
                fig_v.legend(
                    handles=[
                        Line2D([0], [0], marker='o', color='#2ca02c', linestyle='none',
                               markersize=7, label='High viability'),
                        Line2D([0], [0], marker='^', color='#d62728', linestyle='none',
                               markersize=7, label='Low viability'),
                    ],
                    title='Class', fontsize=_FS_LEGEND, title_fontsize=_FS_LEGEND_TITLE,
                    bbox_to_anchor=(0.98, 0.98), loc='upper left', frameon=True
                )
                plt.tight_layout(rect=[0, 0, 0.88, 1])
                _vpath = _save_dir / 'cross_violin_temperature.png'
                fig_v.savefig(_vpath, dpi=300, bbox_inches='tight', facecolor='white')
                plt.show()
                print(f'  ✓ Violin plot saved → {_vpath}')
                _df_v.sort_values('Source', key=lambda s: s.map({'True label': 0, 'NN': 1}), kind='stable').to_csv(_save_dir / 'cross_fig3_violin_data.csv', index=False)
                print(f'  ✓ Fig3 violin data saved → {_save_dir / "cross_fig3_violin_data.csv"}')

            # ── Save graph data to CSVs ─────────────────────────────────────────────────────────
            _rows_m = []
            for _mi, _m in enumerate(_METRIC_LABELS):
                for _fi, _v in enumerate(_NN_DATA[_mi]):
                    _rows_m.append({'metric': _m, 'fold': _fi, 'value': _v})
            pd.DataFrame(_rows_m).to_csv(_save_dir / 'cross_fig1_metrics_data.csv', index=False)
            print(f'  ✓ Fig1 metrics data saved → {_save_dir / "cross_fig1_metrics_data.csv"}')

            _rows_o = []
            for _fi, (_tv, _vv) in enumerate(zip(nn_fold_train_accuracies, nn_fold_accuracies)):
                _rows_o.append({'fold': _fi, 'train_accuracy': _tv, 'val_accuracy': _vv})
            pd.DataFrame(_rows_o).to_csv(_save_dir / 'cross_fig2_overfitting_data.csv', index=False)
            print(f'  ✓ Fig2 overfitting data saved → {_save_dir / "cross_fig2_overfitting_data.csv"}')

    _run_btn.on_click(_do_plots_cross)
    _do_plots_cross()  # run immediately with default values

else:
    import warnings
    from sklearn.model_selection import StratifiedKFold
    from sklearn.preprocessing import StandardScaler
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
    import tensorflow as tf
    from tensorflow import keras

    _all_results = {}   # {cell_line: {metric: [fold_values], 'temperatures': ..., ...}}

    for _cl_name, _cl_csv in CELL_LINE_CSVS.items():
        print(f"\n{'='*60}")
        print(f"  Cell line: {_cl_name}  |  CSV: {Path(_cl_csv).name}")
        print(f"{'='*60}")

        # ── 1. Load & filter ──────────────────────────────────────────────────────
        try:
            _df_raw = pd.read_csv(_cl_csv, sep=None, engine="python", index_col=0)
        except Exception as e:
            print(f"  ERROR loading {_cl_csv}: {e}  → skipping.")
            continue

        _df = _filter_data(_df_raw, label=_cl_name)
        if len(_df) == 0:
            print(f"  No valid rows for {_cl_name} → skipping.")
            continue

        # ── 2. Feature extraction & engineering ───────────────────────────────────────
        _X, _ = extract_features(_df, label=_cl_name)
        if len(_X) < len(_df):
            _df = _df.loc[_X.index]
        _X = engineer_features(_X)

        # ── 3. Feature selection (mirrors config) ─────────────────────────────────
        _y = (_df['Label_day2'].str.strip() == 'Alive').astype(int)
        if MAX_TEMP_COL in _df.columns:
            _temps = _df[MAX_TEMP_COL].values.astype(float)
        else:
            print(f"  WARNING: '{MAX_TEMP_COL}' not found in {_cl_name} → temperature violin will be skipped.")
            _temps = np.full(len(_df), np.nan)

        if USE_FEATURE_SELECTION:
            from sklearn.feature_selection import SelectKBest, mutual_info_classif
            _sel = SelectKBest(mutual_info_classif, k=min(N_FEATURES, _X.shape[1]))
            _X = pd.DataFrame(_sel.fit_transform(_X, _y), index=_X.index,
                               columns=[_X.columns[i] for i in _sel.get_support(indices=True)])

        # ── 4. K-fold cross-validation (RF + NN) ────────────────────────────────
        _skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

        _rf_accs,  _rf_precs,  _rf_recs,  _rf_f1s,  _rf_aucs  = [], [], [], [], []
        _nn_accs,  _nn_precs,  _nn_recs,  _nn_f1s,  _nn_aucs  = [], [], [], [], []
        _rf_tr_accs, _rf_tr_f1s = [], []
        _nn_tr_accs, _nn_tr_f1s = [], []
        _nn_oof_preds = np.full(len(_y), -1, dtype=int)
        _rf_oof_preds = np.full(len(_y), -1, dtype=int)

        for _fold, (_tr_idx, _val_idx) in enumerate(_skf.split(_X, _y), 1):
            _Xtr, _Xval = _X.iloc[_tr_idx], _X.iloc[_val_idx]
            _ytr, _yval = _y.iloc[_tr_idx], _y.iloc[_val_idx]

            _sc = StandardScaler()
            _Xtr_s = _sc.fit_transform(_Xtr)
            _Xval_s = _sc.transform(_Xval)

            # Random Forest
            _rf = RandomForestClassifier(n_estimators=300, max_depth=10,
                                          min_samples_split=5, class_weight='balanced',
                                          random_state=42, n_jobs=-1)
            _rf.fit(_Xtr_s, _ytr)
            _yp_rf       = _rf.predict(_Xval_s)
            _yp_rf_proba = _rf.predict_proba(_Xval_s)[:, 1]
            _rf_oof_preds[_val_idx] = _yp_rf
            _rf_accs.append(accuracy_score(_yval, _yp_rf))
            _rf_precs.append(precision_score(_yval, _yp_rf, zero_division=0))
            _rf_recs.append(recall_score(_yval, _yp_rf, zero_division=0))
            _rf_f1s.append(f1_score(_yval, _yp_rf, zero_division=0))
            try:    _rf_aucs.append(roc_auc_score(_yval, _yp_rf_proba))
            except: _rf_aucs.append(float('nan'))
            _yp_rf_tr = _rf.predict(_Xtr_s)
            _rf_tr_accs.append(accuracy_score(_ytr, _yp_rf_tr))
            _rf_tr_f1s.append(f1_score(_ytr, _yp_rf_tr, zero_division=0))

            # Neural Network
            _ytr_oh = tf.keras.utils.to_categorical(_ytr, 2)
            _nn = keras.Sequential([
                keras.layers.Input(shape=(_Xtr_s.shape[1],)),
                keras.layers.Dense(128, activation='relu',
                                   kernel_regularizer=keras.regularizers.l2(L2_REG)),
                keras.layers.Dropout(DROPOUT_RATE),
                keras.layers.Dense(10, activation='relu',
                                   kernel_regularizer=keras.regularizers.l2(L2_REG)),
                keras.layers.Dropout(DROPOUT_RATE),
                keras.layers.Dense(2, activation='softmax'),
            ])
            _nn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
            _cb = []
            if EARLY_STOPPING:
                _cb.append(keras.callbacks.EarlyStopping(
                    monitor='val_loss', patience=EARLY_STOPPING_PATIENCE,
                    restore_best_weights=True))
            with warnings.catch_warnings():
                warnings.simplefilter('ignore')
                _nn.fit(_Xtr_s, _ytr_oh, epochs=100, batch_size=32,
                        validation_split=0.1, callbacks=_cb, verbose=0)
            _yp_nn_proba = _nn.predict(_Xval_s, verbose=0)[:, 1]
            _yp_nn       = (_yp_nn_proba >= 0.5).astype(int)
            _nn_oof_preds[_val_idx] = _yp_nn
            _nn_accs.append(accuracy_score(_yval, _yp_nn))
            _nn_precs.append(precision_score(_yval, _yp_nn, zero_division=0))
            _nn_recs.append(recall_score(_yval, _yp_nn, zero_division=0))
            _nn_f1s.append(f1_score(_yval, _yp_nn, zero_division=0))
            try:    _nn_aucs.append(roc_auc_score(_yval, _yp_nn_proba))
            except: _nn_aucs.append(float('nan'))
            _yp_nn_tr_proba = _nn.predict(_Xtr_s, verbose=0)[:, 1]
            _yp_nn_tr       = (_yp_nn_tr_proba >= 0.5).astype(int)
            _nn_tr_accs.append(accuracy_score(_ytr, _yp_nn_tr))
            _nn_tr_f1s.append(f1_score(_ytr, _yp_nn_tr, zero_division=0))

            print(f"  Fold {_fold}: RF acc={_rf_accs[-1]:.3f} (train {_rf_tr_accs[-1]:.3f})"
                  f"  NN acc={_nn_accs[-1]:.3f} (train {_nn_tr_accs[-1]:.3f})")

        _all_results[_cl_name] = {
            "RF_Accuracy":  _rf_accs,  "RF_Precision": _rf_precs,
            "RF_Recall":    _rf_recs,  "RF_F1":        _rf_f1s,   "RF_AUC": _rf_aucs,
            "NN_Accuracy":  _nn_accs,  "NN_Precision": _nn_precs,
            "NN_Recall":    _nn_recs,  "NN_F1":        _nn_f1s,   "NN_AUC": _nn_aucs,
            "RF_train_Accuracy": _rf_tr_accs, "RF_train_F1": _rf_tr_f1s,
            "NN_train_Accuracy": _nn_tr_accs, "NN_train_F1": _nn_tr_f1s,
            "temperatures":   _temps,
            "true_labels":    _y.values,
            "rf_oof_preds":   _rf_oof_preds,
            "nn_oof_preds":   _nn_oof_preds,
        }
        _rf_gap = np.nanmean(_rf_tr_accs) - np.nanmean(_rf_accs)
        _nn_gap = np.nanmean(_nn_tr_accs) - np.nanmean(_nn_accs)
        print(f"  ✓ RF: val={np.nanmean(_rf_accs):.3f}  train={np.nanmean(_rf_tr_accs):.3f}  gap={_rf_gap:+.3f}")
        print(f"  ✓ NN: val={np.nanmean(_nn_accs):.3f}  train={np.nanmean(_nn_tr_accs):.3f}  gap={_nn_gap:+.3f}")

    if len(_all_results) == 0:
        print("No cell lines were processed successfully.")
    else:
        import ipywidgets as widgets
        import matplotlib
        from IPython.display import display as _ipy_display

        _cell_lines  = list(_all_results.keys())
        _n_cl        = len(_cell_lines)
        _metrics     = ["Accuracy", "Precision", "Recall", "F1", "AUC"]
        _n_metrics   = len(_metrics)
        _group_width = _n_cl + 1
        _rng         = np.random.default_rng(42)
        _save_dir    = Path(list(CELL_LINE_CSVS.values())[0]).parent

        # ── Summary table ──────────────────────────────────────────────────────────────────────
        print(f"\n{'Cell Line':<15} {'RF val':>8} {'RF train':>9} {'RF gap':>8}"
              f" {'NN val':>8} {'NN train':>9} {'NN gap':>8}")
        print("-" * 75)
        for _cl in _cell_lines:
            r = _all_results[_cl]
            _rf_v = np.nanmean(r['RF_Accuracy']); _rf_t = np.nanmean(r['RF_train_Accuracy'])
            _nn_v = np.nanmean(r['NN_Accuracy']); _nn_t = np.nanmean(r['NN_train_Accuracy'])
            print(f"{_cl:<15} {_rf_v:>8.3f} {_rf_t:>9.3f} {_rf_t-_rf_v:>+8.3f}"
                  f" {_nn_v:>8.3f} {_nn_t:>9.3f} {_nn_t-_nn_v:>+8.3f}")

        # ── Interactive widgets ──────────────────────────────────────────────────────────
        _default_hex = [matplotlib.colors.to_hex(plt.cm.Set2(v))
                        for v in np.linspace(0, 0.9, _n_cl)]
        _color_pickers = {
            cl: widgets.ColorPicker(
                description=cl,
                value=_default_hex[ci],
                style={'description_width': 'initial'},
                layout=widgets.Layout(width='280px')
            )
            for ci, cl in enumerate(_cell_lines)
        }
        _dead_alpha_slider = widgets.FloatSlider(
            value=0.55, min=0.0, max=1.0, step=0.05,
            description='Dead α:',
            style={'description_width': 'initial'},
            layout=widgets.Layout(width='380px')
        )
        _alive_alpha_slider = widgets.FloatSlider(
            value=0.25, min=0.0, max=1.0, step=0.05,
            description='Alive α:',
            style={'description_width': 'initial'},
            layout=widgets.Layout(width='380px')
        )
        _run_btn = widgets.Button(
            description='▶  Update Plots',
            button_style='primary',
            layout=widgets.Layout(width='180px', margin='10px 0 0 0')
        )
        _out = widgets.Output()
        _ipy_display(widgets.VBox([
            widgets.HTML(
                '<b style="font-size:14px; font-family: Times New Roman">'
                'Select a color for each cell line, then click Update Plots:</b>'
            ),
            widgets.VBox(list(_color_pickers.values())),
            widgets.HTML('<i style="font-size:12px">Violin transparency (violin plot only):</i>'),
            _dead_alpha_slider,
            _alive_alpha_slider,
            _run_btn,
            _out,
        ]))

        # ── Times New Roman font settings ─────────────────────────────────────────────────────
        _tnr_rc = {
            'font.family':           'serif',
            'font.serif':            ['Times New Roman'],
            'axes.titlesize':        _FS_AXES_TITLE,
            'axes.titleweight':      'bold',
            'axes.labelsize':        _FS_LABEL,
            'axes.labelweight':      'bold',
            'xtick.labelsize':       _FS_TICK,
            'ytick.labelsize':       _FS_TICK,
            'legend.fontsize':       _FS_LEGEND,
            'legend.title_fontsize': _FS_LEGEND_TITLE,
        }

        def _do_plots(b=None):
            with _out:
                _out.clear_output(wait=True)
                _colors      = [matplotlib.colors.to_rgba(_color_pickers[cl].value)
                                for cl in _cell_lines]
            _dead_alpha  = _dead_alpha_slider.value
            _alive_alpha = _alive_alpha_slider.value

            with matplotlib.rc_context(_tnr_rc):

                # ── Figure 1a & 1b: RF and NN validation metrics ───────────────────────────
                def _plot_metrics_fig(model, fname):
                    fig, ax = plt.subplots(figsize=(max(10, _n_cl * 3), 6))
                    fig.suptitle(f"K-Fold Validation Metrics — {model}",
                                 fontsize=_FS_SUPTITLE, fontweight='bold')
                    _legend_handles = []
                    for _ci, (_cl, _col) in enumerate(zip(_cell_lines, _colors)):
                        _positions = [_gi * _group_width + _ci for _gi in range(_n_metrics)]
                        _data = [_all_results[_cl][f"{model}_{_m}"] for _m in _metrics]
                        ax.boxplot(_data, positions=_positions, widths=0.6, patch_artist=True,
                                   boxprops=dict(facecolor=(*_col[:3], 0.35), color=tuple(_col[:3])),
                                   medianprops=dict(color=tuple(_col[:3]), linewidth=2.5),
                                   whiskerprops=dict(color=tuple(_col[:3]), linewidth=1.2),
                                   capprops=dict(color=tuple(_col[:3]), linewidth=1.2),
                                   flierprops=dict(marker='', markersize=0))
                        for _pos, _vals in zip(_positions, _data):
                            _clean = [v for v in _vals if not np.isnan(v)]
                            _jitter = _rng.uniform(-0.18, 0.18, len(_clean))
                            ax.scatter(np.array(_pos) + _jitter, _clean, color=_col,
                                       s=35, zorder=5, alpha=0.85,
                                       edgecolors='white', linewidths=0.5)
                        _legend_handles.append(plt.Rectangle((0, 0), 1, 1,
                            facecolor=(*_col[:3], 0.55),
                            edgecolor=tuple(_col[:3]), label=_cl))
                    _xtick_pos = [_gi * _group_width + (_n_cl - 1) / 2
                                  for _gi in range(_n_metrics)]
                    ax.set_xticks(_xtick_pos)
                    ax.set_xticklabels(_metrics, fontsize=_FS_TICK, fontweight='bold')
                    ax.set_xlim(-0.8, _n_metrics * _group_width - 1.2)
                    ax.set_ylim(0, 1.08)
                    ax.set_ylabel("Score", fontsize=_FS_LABEL, fontweight='bold')
                    ax.axhline(0.5, color='gray', linestyle='--', linewidth=0.8, alpha=0.5)
                    ax.legend(handles=_legend_handles, title="Cell Line",
                              fontsize=_FS_LEGEND, title_fontsize=_FS_LEGEND_TITLE, bbox_to_anchor=(1.05, 1), loc='upper left',
                                 frameon=True)
                    ax.grid(axis='y', alpha=0.3)
                    ax.spines[['top', 'right']].set_visible(False)
                    ax.tick_params(axis='both', labelsize=_FS_TICK)
                    for _gi in range(1, _n_metrics):
                        ax.axvline(_gi * _group_width - 0.5, color='lightgray', linewidth=0.8)
                    plt.tight_layout()
                    fig.savefig(_save_dir / fname, dpi=300, bbox_inches='tight', facecolor='white')
                    plt.show()

                for _model in ['RF', 'NN']:
                    print(f'── {_model} Mean K-Fold Metrics ──')
                    _header = f"{'Cell Line':<15}" + ''.join(f"{_m:>11}" for _m in _metrics)
                    print(_header)
                    print('-' * (15 + 11 * len(_metrics)))
                    for _cl in _cell_lines:
                        _row = f'{_cl:<15}'
                        for _m in _metrics:
                            _row += f"{np.nanmean(_all_results[_cl][f'{_model}_{_m}']):>11.3f}"
                        print(_row)

                _plot_metrics_fig("RF", "cellline_comparison_metrics_RF.png")
                _plot_metrics_fig("NN", "cellline_comparison_metrics_NN.png")

                # ── Figure 2: overfitting — train vs val accuracy ─────────────────────────
                fig2, axes2 = plt.subplots(1, 2, figsize=(max(10, _n_cl * 2.5), 5), sharey=True)
                fig2.suptitle("Overfitting Analysis: Train vs Validation Accuracy per Fold",
                              fontsize=_FS_SUPTITLE, fontweight='bold')
                _tr_offset = -0.18;  _val_offset = 0.18
                for _mi, _model in enumerate(["RF", "NN"]):
                    ax = axes2[_mi]
                    for _ci, (_cl, _col) in enumerate(zip(_cell_lines, _colors)):
                        _tr_vals  = _all_results[_cl][f"{_model}_train_Accuracy"]
                        _val_vals = _all_results[_cl][f"{_model}_Accuracy"]
                        for _vals, _off, _alpha, _lw in [
                            (_tr_vals,  _tr_offset,  0.55, 2.0),
                            (_val_vals, _val_offset, 0.25, 1.5),
                        ]:
                            ax.boxplot([_vals], positions=[_ci + _off], widths=0.22,
                                       patch_artist=True,
                                       boxprops=dict(facecolor=(*_col[:3], _alpha),
                                                     color=tuple(_col[:3])),
                                       medianprops=dict(color=tuple(_col[:3]), linewidth=_lw),
                                       whiskerprops=dict(color=tuple(_col[:3]), linewidth=1.0),
                                       capprops=dict(color=tuple(_col[:3]), linewidth=1.0),
                                       flierprops=dict(marker='', markersize=0))
                        for _tv, _vv in zip(_tr_vals, _val_vals):
                            ax.plot([_ci + _tr_offset, _ci + _val_offset], [_tv, _vv],
                                    color=tuple(_col[:3]), alpha=0.35, linewidth=0.9, zorder=3)
                        for _vals, _off in [(_tr_vals, _tr_offset), (_val_vals, _val_offset)]:
                            _jit = _rng.uniform(-0.06, 0.06, len(_vals))
                            ax.scatter(np.array(_ci + _off) + _jit, _vals, color=_col,
                                       s=30, zorder=5, alpha=0.9,
                                       edgecolors='white', linewidths=0.4)
                    ax.set_xticks(np.arange(_n_cl))
                    ax.set_xticklabels(_cell_lines, fontsize=_FS_TICK, rotation=15, ha='right')
                    ax.set_ylim(0, 1.08)
                    ax.set_ylabel("Accuracy", fontsize=_FS_LABEL, fontweight='bold')
                    ax.set_title(f"{_model} — Train vs Validation",
                                 fontsize=_FS_AXES_TITLE, fontweight='bold')
                    ax.axhline(0.5, color='gray', linestyle='--', linewidth=0.8, alpha=0.5)
                    ax.grid(axis='y', alpha=0.3)
                    ax.spines[['top', 'right']].set_visible(False)
                    ax.tick_params(axis='both', labelsize=_FS_TICK)
                    from matplotlib.patches import Patch
                    ax.legend(handles=[Patch(facecolor='gray', alpha=0.6, label='Train'),
                                       Patch(facecolor='gray', alpha=0.2, label='Validation')],
                              fontsize=_FS_LEGEND, title_fontsize=_FS_LEGEND_TITLE, loc='lower right')
                    for _ci, _cl in enumerate(_cell_lines):
                        _gap = (np.nanmean(_all_results[_cl][f"{_model}_train_Accuracy"])
                                - np.nanmean(_all_results[_cl][f"{_model}_Accuracy"]))
                        ax.text(_ci, 0.02, f"gap\n{_gap:+.2f}", ha='center', va='bottom',
                                fontsize=_FS_GAP_TEXT, color='#555555', style='italic',
                                fontfamily='Times New Roman')
                plt.tight_layout()
                fig2.savefig(_save_dir / "cellline_overfitting_analysis.png",
                             dpi=300, bbox_inches='tight', facecolor='white')
                plt.show()

                # ── Figure 3: violin — temperature by class ──────────────────────────────────
                _cl_palette   = {cl: tuple(_colors[ci][:3]) for ci, cl in enumerate(_cell_lines)}
                _source_order = ['True label', 'NN']

                _rows_v = []
                for _cl in _cell_lines:
                    _t     = _all_results[_cl]['temperatures']
                    _tl    = _all_results[_cl]['true_labels']
                    _rf_pl = _all_results[_cl]['rf_oof_preds']
                    _nn_pl = _all_results[_cl]['nn_oof_preds']
                    if np.all(np.isnan(_t)):
                        print(f"  Skipping {_cl} in violin plot — no temperature data.")
                        continue
                    for _temp, _true, _rf_p, _nn_p in zip(_t, _tl, _rf_pl, _nn_pl):
                        if not np.isnan(_temp):
                            _rows_v.append({'Cell_Line': _cl, 'Temperature': _temp,
                                            'Source': 'True label',
                                            'Class': 'Alive' if _true == 1 else 'Dead'})
                            _rows_v.append({'Cell_Line': _cl, 'Temperature': _temp,
                                            'Source': 'NN',
                                            'Class': 'Alive' if _nn_p == 1 else 'Dead'})

                if len(_rows_v) == 0:
                    print(f"  WARNING: '{MAX_TEMP_COL}' not found in any cell line"
                          f" — skipping violin plot.")
                else:
                    _df_v = pd.DataFrame(_rows_v)

                    fig_v, axes_v = plt.subplots(
                        1, _n_cl, figsize=(max(14, _n_cl * 3), 7), sharey=True)
                    axes_v = np.atleast_1d(axes_v)
                    fig_v.suptitle(
                        'Temperature Distribution — NN OOF Predicted vs True Label',
                        fontsize=_FS_SUPTITLE, fontweight='bold')

                    for _ci, _cl in enumerate(_cell_lines):
                        ax_v = axes_v[_ci]
                        _df_cl = _df_v[_df_v['Cell_Line'] == _cl]
                        _col = _cl_palette[_cl]

                        sns.violinplot(
                            data=_df_cl[_df_cl['Class'] == 'Dead'],
                            x='Source', y='Temperature', ax=ax_v,
                            order=_source_order, color='#d62728',
                            inner=None, linewidth=3, dodge=False, width=0.8
                        )
                        for artist in ax_v.collections:
                            artist.set_alpha(_dead_alpha)
                            artist.set_linestyle('--')
                        sns.violinplot(
                            data=_df_cl[_df_cl['Class'] == 'Alive'],
                            x='Source', y='Temperature', ax=ax_v,
                            order=_source_order, color='#2ca02c',
                            inner=None, linewidth=3, dodge=False, width=0.8
                        )
                        for artist in ax_v.collections:
                            artist.set_alpha(_alive_alpha)
                        sns.stripplot(
                            data=_df_cl[_df_cl['Class'] == 'Dead'],
                            x='Source', y='Temperature', ax=ax_v,
                            order=_source_order, dodge=True, jitter=True,
                            alpha=0.85, size=7, marker='^', color='#d62728',
                            linewidth=0.3, edgecolor='white', legend=False
                        )
                        sns.stripplot(
                            data=_df_cl[_df_cl['Class'] == 'Alive'],
                            x='Source', y='Temperature', ax=ax_v,
                            order=_source_order, dodge=True, jitter=True,
                            alpha=0.4, size=7, marker='o', color='#2ca02c',
                            linewidth=0.3, edgecolor='white', legend=False
                        )

                        ax_v.set_title(_cl, fontsize=_FS_AXES_TITLE, fontweight='bold')
                        ax_v.set_xlabel('Prediction source', fontsize=_FS_LABEL, fontweight='bold')
                        if _ci == 0:
                            ax_v.set_ylabel('Temperature', fontsize=_FS_LABEL, fontweight='bold')
                        else:
                            ax_v.set_ylabel('')
                        ax_v.set_ylim(20, 100)
                        ax_v.grid(True, alpha=0.3, axis='y')
                        ax_v.tick_params(axis='both', labelsize=_FS_TICK)
                        if ax_v.get_legend() is not None:
                            ax_v.get_legend().remove()

                    from matplotlib.lines import Line2D
                    _mk_leg = [
                        Line2D([0], [0], color='#2ca02c', linestyle='-',  linewidth=1.5,
                               marker='o', markersize=7, label='High viability'),
                        Line2D([0], [0], color='#d62728', linestyle='--', linewidth=1.5,
                               marker='^', markersize=7, label='Low viability'),
                    ]
                    fig_v.subplots_adjust(right=0.82)
                    fig_v.legend(handles=_mk_leg, title='Class',
                                 fontsize=_FS_LEGEND, title_fontsize=_FS_LEGEND_TITLE,
                                 bbox_to_anchor=(0.98, 0.98), loc='upper left',
                                 frameon=True)
                    plt.tight_layout(rect=[0, 0, 0.82, 1])
                    _vpath = _save_dir / 'cellline_violin_temperature.png'
                    fig_v.savefig(_vpath, dpi=300, bbox_inches='tight', facecolor='white')
                    plt.show()
                    print(f'  ✓ Violin plot saved → {_vpath}')
                    _df_v.sort_values('Source', key=lambda s: s.map({'True label': 0, 'NN': 1}), kind='stable').to_csv(_save_dir / 'fig3_violin_data.csv', index=False)
                    print(f'  ✓ Fig3 violin data saved → {_save_dir / "fig3_violin_data.csv"}')

                # ── Save graph data to CSVs ─────────────────────────────────────────────────────────
                _rows_m = []
                for _cl in _cell_lines:
                    for _m in _metrics:
                        for _fi, _v in enumerate(_all_results[_cl][f'NN_{_m}']):
                            _rows_m.append({'cell_line': _cl, 'metric': _m,
                                            'fold': _fi, 'value': _v})
                pd.DataFrame(_rows_m).to_csv(_save_dir / 'fig1_metrics_data.csv', index=False)
                print(f'  ✓ Fig1 metrics data saved → {_save_dir / "fig1_metrics_data.csv"}')

                _rows_o = []
                for _cl in _cell_lines:
                    _tr = _all_results[_cl]['NN_train_Accuracy']
                    _va = _all_results[_cl]['NN_Accuracy']
                    for _fi, (_tv, _vv) in enumerate(zip(_tr, _va)):
                        _rows_o.append({'cell_line': _cl, 'fold': _fi,
                                        'train_accuracy': _tv, 'val_accuracy': _vv})
                pd.DataFrame(_rows_o).to_csv(_save_dir / 'fig2_overfitting_data.csv', index=False)
                print(f'  ✓ Fig2 overfitting data saved → {_save_dir / "fig2_overfitting_data.csv"}')

        _run_btn.on_click(_do_plots)
        _do_plots()  # run immediately with default values
```
